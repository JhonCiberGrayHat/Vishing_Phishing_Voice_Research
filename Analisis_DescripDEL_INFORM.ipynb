```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para gráficos
sns.set(style="whitegrid")
```

```python
#Cargar el dataset
df = pd.read_csv("Conteo_de_Víctimas_V2_20251211_delitos_informaticos.csv")

# Exploración inicial
print("Dimensiones del dataset:", df.shape)
df.head()
```

```python
# Información general
df.info()

# Estadísticas descriptivas
df.describe(include="all")
```

```python
# Conteo de valores nulos por columna
df.isnull().sum()
```

```python
#Distribución de variables categóricas
for col in df.select_dtypes(include="object").columns:
    print(f"\nDistribución de {col}:")
    print(df[col].value_counts())
```

```python
#Visualizaciones básicas

# Seleccionar columnas categóricas
categorical_cols = df.select_dtypes(include="object").columns

# Generar gráficas para cada variable categórica
for col in categorical_cols:
    plt.figure(figsize=(10,6))
    sns.countplot(data=df, x=col, order=df[col].value_counts().index)
    plt.xticks(rotation=45)
    plt.title(f"Distribución de {col}")
    plt.xlabel(col)
    plt.ylabel("Frecuencia")
    plt.show()
```

```python
#Relación entre variables numéricas

num_cols = df.select_dtypes(include="number").columns
if len(num_cols) >= 2:
    plt.figure(figsize=(8,6))
    sns.scatterplot(data=df, x=num_cols[0], y=num_cols[1], hue=df.columns[0])
    plt.title(f"Relación entre {num_cols[0]} y {num_cols[1]}")
    plt.show()
```

```python

# Correlaciones

import numpy as np

plt.figure(figsize=(10,8))
# Select only numeric columns for correlation calculation
numeric_df = df.select_dtypes(include=[np.number])  # Only include numeric data types
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Mapa de correlaciones")
plt.show()
```

```python
#Filtrar por los delitos específicos
# Lista de delitos a filtrar


delitos_filtrar = [
    "HURTO POR MEDIOS INFORMATICOS Y SEMEJANTES ART. 269I LEY 1273 DE 2009",
    "VIOLACION DE DATOS PERSONALES ART 269F LEY 1273 DE 2009",
    "ACCESO ABUSIVO A UN SISTEMA INFORMATICO ART 269A LEY 1273 DE 2009",
    "TRANSFERENCIA NO CONSENTIDA DE ACTIVOS VALIENDOSE DE ALGUNA MANIPULACION INFORMATICA O ARTIFICIO SEMEJANTE ART. 269J LEY 1273 DE 2009"
]

# Filtrar el dataset
df_filtrado = df[df['DELITO'].isin(delitos_filtrar)]

# Verificar resultado
df_filtrado['DELITO'].value_counts()
```

```python
#Graficar cada delito de manera independiente

for delito in delitos_filtrar:
    datos_delito = df_filtrado[df_filtrado['DELITO'] == delito]
    
    plt.figure(figsize=(10,6))
    sns.countplot(data=datos_delito, x='DELITO')
    plt.title(f"Distribución de casos: {delito}")
    plt.xlabel("Delito")
    plt.ylabel("Frecuencia")
    plt.xticks(rotation=45)
    plt.show()
```

```python
# Crear columna auxiliar para clasificar delitos
# Lista de delitos específicos
delitos_especificos = [
    "HURTO POR MEDIOS INFORMATICOS Y SEMEJANTES ART. 269I LEY 1273 DE 2009",
    "VIOLACION DE DATOS PERSONALES ART 269F LEY 1273 DE 2009",
    "ACCESO ABUSIVO A UN SISTEMA INFORMATICO ART 269A LEY 1273 DE 2009",
    "TRANSFERENCIA NO CONSENTIDA DE ACTIVOS VALIENDOSE DE ALGUNA MANIPULACION INFORMATICA O ARTIFICIO SEMEJANTE ART. 269J LEY 1273 DE 2009"
]


df['Categoria'] = df['DELITO'].apply(lambda x: x if x in delitos_especificos else "Otros delitos informáticos")

# Conteo de casos
conteo = df['Categoria'].value_counts()
total = conteo.sum()

# Gráfica comparativa
plt.figure(figsize=(20,10))
ax = sns.barplot(x=conteo.index, y=conteo.values, palette="viridis")

# Añadir etiquetas con número y porcentaje sobre cada barra
for i, valor in enumerate(conteo.values):
    porcentaje = (valor / total) * 100
    ax.text(i, valor + 5, f"{valor} ({porcentaje:.1f}%)", 
            ha='center', va='bottom', fontsize=20, fontweight='bold')

plt.title("Comparación de delitos informáticos")
plt.xlabel("Categoría de delito")
plt.ylabel("Número de casos")
plt.xticks(rotation=45)
plt.show()
```
